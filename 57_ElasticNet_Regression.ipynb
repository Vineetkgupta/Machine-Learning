```python
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression, Lasso, Ridge, ElasticNet
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
X,y = load_diabetes(return_X_y=True)
```

```python
# train - test split 
X_train, X_test, y_train, y_test = train_test_split(X,y, random_state=2, test_size=0.2)
```

```python
# Linear Regression
lr = LinearRegression()
lr.fit(X_train, y_train)
y_pred = lr.predict(X_test)
r2_score(y_test, y_pred)
```

```python
# Ridge Regression
rr = Ridge(alpha=0.1)
rr.fit(X_train, y_train)
y_pred = rr.predict(X_test)
r2_score(y_test, y_pred)
```

```python
# Lasso Regression
la = Lasso(alpha=0.1)
la.fit(X_train, y_train)
y_pred = la.predict(X_test)
r2_score(y_test, y_pred) # In this case, Ridge is performing better than Lasso
```

```python
# ElasticNet
en = ElasticNet(alpha=0.005, l1_ratio=0.9) # l1_ratio = 0.9 means we are giving more weighted to Ridge as compared to Lasso
en.fit(X_train, y_train)
y_pred = en.predict(X_test)
r2_score(y_test, y_pred) # we are getting slightly better result in this case
```
