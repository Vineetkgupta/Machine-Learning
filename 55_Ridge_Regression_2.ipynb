```python
from sklearn.datasets import make_regression
import numpy as np
import pandas as pd
```

```python
X,y = make_regression(n_samples = 100, n_features = 1, n_informative = 1, n_targets=1, noise = 20, random_state =13)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(X,y)
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(X,y)
```

```python
print(lr.coef_)
print(lr.intercept_)
```

```python
from sklearn.linear_model import Ridge
```

```python
rr1 = Ridge(alpha=10) 
rr1.fit(X,y)
print(rr1.coef_)
print(rr1.intercept_)
# as we can see here after setting alpha =10, coeff comes down to 24
```

```python
rr2 = Ridge(alpha=100) # alpha = 100
rr2.fit(X,y)
print(rr2.coef_)
print(rr2.intercept_)
```

```python
plt.plot(X,y,'b.')
plt.plot(X, lr.predict(X), color = 'red', label = 'alpha=0')
plt.plot(X, rr1.predict(X), color = 'green', label = 'alpha=0')
plt.plot(X, rr2.predict(X), color = 'yellow', label = 'alpha =100')
# we can see here when as we increases the values of alpha then slop decreases
```
