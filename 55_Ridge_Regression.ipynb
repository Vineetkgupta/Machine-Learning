```python
# Ridge Regression(L2 Regularization) is used toshrinks coefficient to reduce overfitting (Overfitting- Performing very well in training data
# and very bad on testing data)
# Suppose, we have equation of line y = mx + b , if m is much large then it means y is much dependent of x causes overfitting.
#                                                if m is much small then it means y is less dependent of x causes underfitting.
  
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import load_diabetes
data = load_diabetes()
```

```python
print(data.DESCR)
```

```python
X = data.data
y = data.target
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2, random_state=45)
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(X_train, y_train)
```

```python
y_pred = lr.predict(X_test)
```

```python
from sklearn.metrics import r2_score, root_mean_squared_error
```

```python
print("R2_score: ", r2_score(y_test,y_pred))
```

```python
print("root_mean_squared_error: ",root_mean_squared_error(y_test,y_pred))
```

```python
from sklearn.linear_model import Ridge
R = Ridge(alpha = 000.1)
```

```python
R.fit(X_train, y_train)
```

```python
y_pred = R.predict(X_test)
```

```python
print("r2_Score: ", r2_score(y_test, y_pred))
print("root_mean_squared_error: ", root_mean_squared_error(y_test,y_pred)) # there is improvement but not significant but used in most of the models
```
